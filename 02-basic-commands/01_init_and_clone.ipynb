## Introduction: Two Ways to Start

There are two main ways to start working with Git:

1. **`git init`** - Create a brand new repository from scratch
2. **`git clone`** - Copy an existing repository from GitHub (or elsewhere)

Think of it like this:
- `git init` = Starting a new research project
- `git clone` = Joining an existing research project

## Part 1: Creating a New Repository with `git init`

````python
# Create a new directory for our project
!mkdir -p /tmp/my-first-repo
!cd /tmp/my-first-repo && pwd
````

````python
# Initialize a Git repository
!cd /tmp/my-first-repo && git init
````

**What just happened?**

Git created a hidden `.git` directory that contains all the version control information. This is where Git stores:
- All commits (snapshots of your code)
- Branch information
- Configuration settings
- History metadata

````python
# Look at what Git created
!cd /tmp/my-first-repo && ls -la
````

````python
# Peek inside the .git directory
!cd /tmp/my-first-repo && ls -la .git/
````

## Understanding the .git Directory

The `.git` directory is the heart of your repository:

```
.git/
├── HEAD              # Points to current branch
├── config            # Repository configuration
├── description       # Repository description
├── hooks/            # Scripts that run at certain events
├── objects/          # All content (commits, files, etc.)
├── refs/             # Pointers to commits (branches, tags)
└── index             # Staging area
```

⚠️ **Important:** Never manually edit files in `.git/` unless you know what you're doing!

````python
# Check the status of our new repository
!cd /tmp/my-first-repo && git status
````

## Part 2: Cloning an Existing Repository

More commonly, you'll want to work with existing repositories. Let's clone one!

````python
# Clone a sample repository (this tutorial!)
!cd /tmp && git clone https://github.com/melhzy/github_tutorial.git
````

**What just happened?**

Git:
1. Created a new directory named `github_tutorial`
2. Initialized a `.git` directory inside it
3. Downloaded all the repository's history
4. Checked out the default branch (usually `main` or `master`)
5. Set up a connection to the original repository (called `origin`)

````python
# Explore what was cloned
!cd /tmp/github_tutorial && ls -la
````

````python
# Check the repository status
!cd /tmp/github_tutorial && git status
````

## Clone Options and Variations

### Basic Clone
```bash
git clone <url>
```

### Clone to a Specific Directory
```bash
git clone <url> my-custom-name
```

### Clone a Specific Branch
```bash
git clone -b branch-name <url>
```

### Shallow Clone (Faster, Less History)
```bash
git clone --depth 1 <url>
```
Useful for large repositories when you don't need full history.

````python
# Example: Clone to a custom directory name
!cd /tmp && git clone https://github.com/melhzy/github_tutorial.git my-tutorial
````

## Understanding Remote Connections

When you clone, Git automatically sets up a "remote" connection:

````python
# View remote connections
!cd /tmp/github_tutorial && git remote -v
````

**Output explanation:**
- `origin` - Default name for the remote repository
- `fetch` - URL for downloading updates
- `push` - URL for uploading changes

These are usually the same URL.

## Init vs Clone: When to Use What?

### Use `git init` when:
- Starting a completely new project
- Converting an existing local project to use Git
- You want to create something from scratch

### Use `git clone` when:
- Downloading someone else's project
- Getting your own project on a new machine
- Contributing to open source
- Working with teammates on shared projects

**In research:** You'll use clone much more often!

## Common Issues and Solutions

### "fatal: destination path already exists"
- The directory already exists
- Solution: Use a different name or delete the existing directory

### "Permission denied (publickey)"
- SSH authentication issue
- Solution: Use HTTPS URL or set up SSH keys (see Module 01.3)

### "Repository not found"
- Wrong URL or repository is private
- Solution: Check URL and your access permissions
