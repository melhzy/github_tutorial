## Introduction: Why View History?

One of Git's superpowers is the ability to travel through time:
- See what changed and when
- Find when bugs were introduced
- Understand project evolution
- Restore previous versions
- Review team contributions

## Basic `git log`

````python
# View commit history
!cd /tmp/github_tutorial && git log
````

## Useful Log Formats

````python
# One line per commit
!cd /tmp/github_tutorial && git log --oneline
````

````python
# Show last 5 commits with statistics
!cd /tmp/github_tutorial && git log --oneline --stat -5
````

````python
# Graphical representation of branches
!cd /tmp/github_tutorial && git log --oneline --graph --all --decorate
````

## Viewing Changes with `git diff`

````python
# View changes in a specific file
!cd /tmp/github_tutorial && git log --oneline -- README.md
````

## Useful Log Options

### Filtering by Author
```bash
git log --author="John"
```

### Filtering by Date
```bash
git log --since="2 weeks ago"
git log --after="2024-01-01"
```

### Filtering by Message
```bash
git log --grep="bug fix"
```

### Show Specific File History
```bash
git log -- path/to/file
```

## Key Takeaways

✅ `git log` shows commit history

✅ Use `--oneline` for compact view

✅ Use `--graph` to visualize branches

✅ `git diff` shows changes between commits

✅ Many options to filter and format log output
