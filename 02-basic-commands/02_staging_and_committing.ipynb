## The Three-Stage Workflow

Git has three main states for your files:

```
Working Directory  →  Staging Area  →  Repository (.git)
  (Modified)          (git add)        (git commit)
```

1. **Working Directory**: Where you edit files
2. **Staging Area**: Where you prepare the next commit
3. **Repository**: Where Git permanently stores snapshots

**Analogy:** Taking a group photo
- Working Directory = Everyone in the room
- Staging Area = People arranged in front of the camera
- Repository = The actual photograph taken

## Creating and Tracking Files

````python
# Create a new file
!cd /tmp/practice-repo && echo "# My Research Project" > README.md
````

````python
# Check status - see the untracked file
!cd /tmp/practice-repo && git status
````

## Staging Files with `git add`

````python
# Stage the file
!cd /tmp/practice-repo && git add README.md
````

````python
# Check status again - file is now staged
!cd /tmp/practice-repo && git status
````

## Committing Changes

A commit is a snapshot of your project at a specific point in time.

````python
# Configure Git (if not done already)
!cd /tmp/practice-repo && git config user.email "you@example.com"
!cd /tmp/practice-repo && git config user.name "Your Name"
````

````python
# Make a commit
!cd /tmp/practice-repo && git commit -m "Initial commit: Add README"
````

## Writing Good Commit Messages

### Anatomy of a Good Commit Message

```
Short summary (50 characters or less)

More detailed explanation if needed. Wrap at 72 characters.
Explain what and why, not how.

- You can use bullet points
- To list multiple changes
```

### Best Practices
- Use imperative mood: "Add feature" not "Added feature"
- Be concise but descriptive
- Reference issues if applicable: "Fix #123"

**Good Examples:**
- `Add data preprocessing pipeline`
- `Fix memory leak in training loop`
- `Update dependencies to latest versions`

**Bad Examples:**
- `fixed stuff`
- `asdf`
- `WIP` (without context)

## Key Takeaways

✅ Git has a three-stage workflow: Working → Staging → Repository

✅ Use `git add` to stage changes

✅ Use `git commit` to save snapshots

✅ Write clear, descriptive commit messages

✅ Commit early and often
