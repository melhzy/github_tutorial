## What are Branches?

Branches are **independent lines of development** in your repository. Think of them as parallel universes for your code:

```
main:     A---B---C---F---G
               \
feature:        D---E
```

### Why Use Branches?

1. **Experimentation**: Try new ideas without breaking working code
2. **Feature Development**: Build new features in isolation
3. **Bug Fixes**: Fix issues without disrupting ongoing work
4. **Collaboration**: Multiple people can work simultaneously
5. **Code Review**: Submit changes for review before merging

## The Default Branch

Every Git repository starts with one branch:
- Modern repos: `main`
- Older repos: `master`

This is your primary, stable branch.

## Branching Strategies

### 1. Feature Branch Workflow
Create a branch for each new feature:
```
main
├── feature/user-auth
├── feature/data-pipeline
└── feature/ml-model
```

### 2. Git Flow
Multiple permanent branches:
- `main`: Production-ready code
- `develop`: Integration branch
- `feature/*`: New features
- `hotfix/*`: Urgent fixes

### 3. GitHub Flow (Simple)
- `main` + feature branches
- Deploy from main
- Perfect for continuous deployment

**For Research:** Feature branch workflow is usually sufficient!

## When to Create a Branch?

### Good Reasons:
- ✅ Adding a new analysis method
- ✅ Experimenting with different models
- ✅ Trying different parameters
- ✅ Fixing a bug
- ✅ Updating documentation

### General Rule:
**One branch = One task/feature/fix**

## Key Takeaways

✅ Branches enable parallel development

✅ Create branches for features, experiments, and fixes

✅ `main` branch should always be stable

✅ Branches are lightweight and fast in Git

✅ Choose a branching strategy that fits your workflow
