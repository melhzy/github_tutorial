## Introduction: Why Bother with Git?

As a researcher, you're constantly creating new versions of your work: code, datasets, figures, and manuscripts. How do you keep track of it all? 

- `analysis_final.py`
- `analysis_final_v2.py`
- `analysis_final_actually_final.py`

If this looks familiar, you're in the right place. **Git** is a **version control system** that solves this problem. It's like having a superpower that lets you track every change, travel back in time to previous versions, and collaborate with others without chaos.

**GitHub** is a web-based platform built around Git, providing a central place to store your code, collaborate, and showcase your work. This tutorial series uses Google Colab, which gives you a direct, hands-on Git experience in your browser.

### Part 1: Checking Your Git Installation

The great news is that Google Colab comes with Git pre-installed! We can verify this by running a simple command. The `!` at the beginning of the line tells Colab to execute this as a shell command.

````python
# Check the installed Git version
!git --version
````

You should see an output like `git version 2.34.1`. The exact version number might differ, but as long as you see a version number, you're all set!

### Part 2: First-Time Git Configuration

Before you start using Git, you need to introduce yourself. Git embeds this information into every change you make (a "commit"), so it's clear who contributed what. This is essential for collaboration.

We only need to do this once per environment. In Colab, this configuration will last for the duration of your session.

Let's configure your **name** and **email address**. 


````python
# In the fields below, replace "Tutorial User" with your name or GitHub username.
# Replace "tutorial@example.com" with the email associated with your GitHub account.
!git config --global user.name "Tutorial User"
!git config --global user.email "tutorial@example.com"
````

These commands don't produce any output, but they save your configuration. The `--global` flag means this setting will apply to every Git project you work on in this environment.

Let's verify that the configuration was saved correctly.

````python
# List all global configurations to check your name and email
!git config --global --list
````

You should see your `user.name` and `user.email` in the output list. Success! ✓

### (Optional) Installing Git on Your Local Machine

While you can follow this entire tutorial in Colab, you'll eventually want Git on your own computer. Here's a quick guide:

- **Linux (Ubuntu/Debian)**:
  ```bash
  sudo apt-get update
  sudo apt-get install git
  ```

- **macOS**:
  The easiest way is to install Xcode Command Line Tools. Open your terminal and run:
  ```bash
  xcode-select --install
  ```
  Alternatively, if you use <a href="https://brew.sh/">Homebrew</a>: `brew install git`

- **Windows**:
  Download and install <a href="https://git-scm.com/download/win">Git for Windows</a>. This package includes Git, Git Bash (a command-line environment), and GUI tools.

### (Optional) A Quick Look at GitHub CLI

<a href="https://cli.github.com/">GitHub CLI</a> (`gh`) is a command-line tool that brings GitHub's features—like pull requests, issues, and releases—to your terminal. It simplifies many common GitHub tasks.

You don't need it for this tutorial, but it's a powerful tool for your toolbox.

**Installation:**
- **Linux (Ubuntu/Debian)**: `sudo apt install gh`
- **macOS (Homebrew)**: `brew install gh`
- **Windows (Winget)**: `winget install --id GitHub.cli`

After installing, you would run `gh auth login` to authenticate with your GitHub account.

## Common Pitfalls &amp; Troubleshooting

- **Using a different email**: Always use the email address associated with your GitHub account for `user.email`. This links your commits to your GitHub profile.
- **Temporary Configuration in Colab**: Remember that your Git configuration in Colab is temporary and will reset when your session disconnects. You'll need to re-run the `git config` cells at the start of each new session.
- **`git: command not found` (Local)**: If you see this on your local machine, it means Git isn't installed or isn't in your system's PATH. Double-check your installation steps.
