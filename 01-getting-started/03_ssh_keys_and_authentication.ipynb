## Introduction: Why Authentication Matters

When you interact with GitHub (pushing code, pulling from private repos), GitHub needs to verify your identity. Think of it like showing your ID card at a secure building.

### Authentication Methods

**1. HTTPS with Personal Access Token (PAT)**
- Uses username and token
- Easier for beginners
- Need to enter token frequently (unless cached)

**2. SSH Keys (Recommended)**
- Uses cryptographic key pairs
- More secure
- No need to enter credentials repeatedly
- Industry standard for developers

**Note:** GitHub deprecated password authentication in 2021 for security reasons.

## Understanding SSH Keys

SSH (Secure Shell) uses **public-key cryptography**:

```
SSH Key Pair
├── Private Key (id_ed25519) - Stay on your computer! Never share!
└── Public Key (id_ed25519.pub) - Share with GitHub
```

### How It Works
1. You generate a key pair (public + private)
2. You add the **public key** to GitHub
3. When connecting, GitHub uses the public key to verify you have the matching private key
4. No passwords sent over the network!

**Analogy:** The public key is like a lock you give to GitHub. Only your private key can unlock it.

## Step 1: Check for Existing SSH Keys

````python
# Check if SSH keys already exist
!ls -la ~/.ssh/
````

If you see files like `id_rsa`, `id_ed25519`, or `id_ecdsa`, you already have SSH keys. Otherwise, we'll generate them in the next step.

**Note:** In Google Colab, you'll need to generate new keys each session since the environment is temporary.

## Step 2: Generate SSH Keys

We'll use the **Ed25519** algorithm, which is modern, secure, and fast.

````python
# Generate a new SSH key
# Replace with your GitHub email address
import os

# Set your email here
email = "your-email@example.com"  # CHANGE THIS!

# Generate the key
!ssh-keygen -t ed25519 -C "{email}" -f ~/.ssh/id_ed25519 -N ""
````

**Command Breakdown:**
- `-t ed25519`: Use Ed25519 algorithm
- `-C "email"`: Add a label (your email)
- `-f ~/.ssh/id_ed25519`: File location
- `-N ""`: No passphrase (for Colab convenience; use passphrase locally!)

## Step 3: Start the SSH Agent

The SSH agent manages your keys in memory.

````python
# Start the SSH agent
!eval "$(ssh-agent -s)"
````

````python
# Add your key to the agent
!ssh-add ~/.ssh/id_ed25519
````

## Step 4: Display Your Public Key

We need to copy the public key to add it to GitHub.

````python
# Display your public key
!cat ~/.ssh/id_ed25519.pub
````

**Copy the entire output** (it should start with `ssh-ed25519` and end with your email). You'll need this in the next step.

⚠️ **Important:** This is your PUBLIC key. Never share your private key (the file without .pub)!

## Step 5: Add SSH Key to GitHub

Now we need to add the public key to your GitHub account:

1. Go to **GitHub.com** and sign in
2. Click your profile picture (top right) → **Settings**
3. In the left sidebar, click **SSH and GPG keys**
4. Click **New SSH key** (green button)
5. Give it a title (e.g., "Google Colab" or "My Laptop")
6. Paste your public key in the "Key" field
7. Click **Add SSH key**
8. Confirm with your GitHub password if prompted

**Direct Link:** https://github.com/settings/keys

## Step 6: Test Your Connection

````python
# Test SSH connection to GitHub
!ssh -T git@github.com
````

**Expected Output:**

If you see this message, **congratulations!** 🎉 Your SSH key is working!

## Understanding SSH vs HTTPS URLs

When cloning or adding remotes, you'll see two URL formats:

**SSH Format (use this now):**
```
git@github.com:username/repository.git
```

**HTTPS Format:**
```
https://github.com/username/repository.git
```

With SSH keys configured, always use the SSH URL for your repositories.

## Personal Access Tokens (Alternative)

If you prefer HTTPS or can't use SSH, you can use Personal Access Tokens:

### Creating a PAT
1. Go to GitHub → Settings → Developer settings
2. Personal access tokens → Tokens (classic)
3. Generate new token
4. Select scopes (repo, workflow, etc.)
5. Generate and **save it securely**
6. Use it as your password when prompted

**Direct Link:** https://github.com/settings/tokens

⚠️ **Warning:** Treat PATs like passwords. Never commit them to your repository!

## Security Best Practices

### Do's ✅
- Generate separate keys for different machines
- Use passphrases for private keys (on personal computers)
- Regularly review and remove unused keys from GitHub
- Keep your private keys secure

### Don'ts ❌
- Never share your private key
- Never commit keys to repositories
- Don't reuse keys across services
- Don't use old algorithms (RSA < 2048 bits)

## Troubleshooting

### "Permission denied (publickey)"
- Check that you copied the entire public key
- Verify the key is added to GitHub
- Make sure ssh-agent is running
- Try `ssh-add -l` to list loaded keys

### "Could not resolve hostname"
- Check your internet connection
- Verify the URL format is correct
