## Local vs Remote Repositories

```
Your Computer (Local)     ←→     GitHub (Remote)
    .git directory               Cloud storage
         ↓                             ↓
    git push  →  Send changes
    git pull  ←  Get changes
```

### Key Concepts
- **Local**: Your computer's copy of the repository
- **Remote**: The GitHub (cloud) copy
- **Origin**: Default name for the remote repository
- **Upstream**: Sometimes refers to the original repository you forked from

## Understanding `git push`

`git push` uploads your local commits to GitHub.

### Basic Syntax
```bash
git push <remote> <branch>
```

### Common Commands
```bash
git push origin main          # Push to main branch
git push origin feature-xyz   # Push to feature branch
git push -u origin main       # Set upstream and push
git push                      # Push to tracked branch
```

## Understanding `git pull`

`git pull` downloads changes from GitHub and merges them.

### What Pull Does
1. Fetches changes from remote
2. Merges them into your current branch

### Basic Syntax
```bash
git pull <remote> <branch>
```

### Common Commands
```bash
git pull origin main          # Pull from main branch
git pull                      # Pull from tracked branch
git pull --rebase            # Pull and rebase instead of merge
```

## Fetch vs Pull

### `git fetch`
- Downloads changes but doesn't merge
- Safe - lets you review before merging
- Updates remote-tracking branches

### `git pull`
- Downloads AND merges automatically
- Convenience command: `fetch` + `merge`
- Can cause merge conflicts

```bash
git pull = git fetch + git merge
```

## Common Workflow

### Daily Research Workflow
```bash
# Start of day: Get latest changes
git pull

# Make changes to your code
# ...

# Stage and commit
git add .
git commit -m "Add new analysis"

# Share your work
git push

# End of day: Push again to backup
git push
```

## Common Scenarios

### Scenario 1: First Push to New Branch
```bash
git checkout -b new-feature
# ... make changes ...
git commit -m "Add feature"
git push -u origin new-feature  # -u sets upstream
```

### Scenario 2: Remote Has Changes
```bash
git push
# Error: remote has changes
git pull                        # Get remote changes first
# Resolve any conflicts
git push                        # Now push succeeds
```

### Scenario 3: Multiple Collaborators
```bash
# Always pull before starting work
git pull
# Work on your changes
# Before pushing, pull again
git pull
git push
```

## Best Practices

### Do's ✅
- Pull before you start working
- Commit and push frequently
- Write descriptive commit messages
- Pull before pushing
- Push at the end of each day (backup!)

### Don'ts ❌
- Don't force push unless you know what you're doing
- Don't push untested code to main
- Don't forget to pull regularly
- Don't push sensitive data (passwords, API keys)

## Key Takeaways

✅ `git push` uploads your commits to GitHub

✅ `git pull` downloads and merges changes from GitHub

✅ `git fetch` downloads without merging

✅ Always pull before pushing

✅ Push regularly to backup your work
